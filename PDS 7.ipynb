A. Write a Python program to create a scatter plot of Age and Height of 5 students.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([15,20,21,25,26])
y = np.array([180,170,145,156,160])

plt.scatter(x, y)
plt.xlabel("Age")
plt.ylabel("Height in cm")
plt.title("Scatter Plot Age and Height")
plt.show()
```

B. Write a Python program to create a bar chart showing marks of 5 students.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["Aniket", "Jay", "Amit", "Kaushal","Raj"])
y = np.array([80, 88, 78, 65,70])

plt.bar(x, y, color = "Blue")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Bar chart Student and Marks")
plt.show()
```

C. Write a Python program to create a histogram showing the distribution of marks of 10 students.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([68, 69, 70, 75, 78, 82, 81, 83, 90])

plt.hist(x)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Histogram of Student Marks")
plt.show()
```

D. Write a Python program to create a pie chart showing the number of
students in different courses.

```python
import matplotlib.pyplot as plt

courses = ["BSc IT", "BCA", "BSc CS", "BBA"]
students = [40, 30, 20, 10]

plt.pie(students, labels=courses, autopct="%1.1f%%")

plt.title("Number of Students in Different Courses")

plt.show()
```

E. Write a Python program to create a bar chart showing sales of 5 different
products.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Keyboard", "Mouse"]
sales = [50, 80, 40, 60, 90]

plt.bar(products, sales)

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales of 5 Different Products")

plt.show()
```
